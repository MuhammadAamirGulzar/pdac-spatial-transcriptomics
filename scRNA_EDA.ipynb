```r
# ==============================================
# EDA Notebook: Filter Seurat object for IU_PDA_HM9
# ==============================================

# --- Install / Load Required Packages ---
required <- c("Seurat", "dplyr", "ggplot2", "patchwork")
missing_pkgs <- required[!sapply(required, requireNamespace, quietly = TRUE)]
if(length(missing_pkgs) > 0){
  stop(paste0("Missing required packages: ", paste(missing_pkgs, collapse = ", "))) 
}

library(Seurat)
library(dplyr)
library(ggplot2)
library(patchwork)
```

```r
# --- Parameters ---
# Input file (change if needed)
infile <- "scRNA-seq_Data_post_qc.rds"

# Output directory for EDA results
outdir <- "outputs"
dir.create(outdir, showWarnings = FALSE, recursive = TRUE)

# Target patient
target <- "IU_PDA_HM9"
```

```r
# --- Read Seurat Object ---
if (!file.exists(infile)) stop(paste0("Input file not found: ", infile))
cat("Reading Seurat object from: ", infile, "\n")
obj <- readRDS(infile)

if (!inherits(obj, "Seurat")) stop("Input object is not a Seurat object.")
```

```r
# --- Identify Metadata Column Containing Patient ---
meta <- obj@meta.data
cols <- colnames(meta)

# Try exact match first
exact_cols <- cols[sapply(cols, function(cn) any(as.character(meta[[cn]]) == target))]
if(length(exact_cols) == 0){
  # fallback: substring match
  fuzzy_cols <- cols[sapply(cols, function(cn) any(grepl(target, as.character(meta[[cn]]), fixed = TRUE)))]
} else {
  fuzzy_cols <- exact_cols
}

if(length(fuzzy_cols) == 0){
  stop(paste0("Could not find any metadata column containing '\
'."))
}

meta_col <- fuzzy_cols[1]
cat("Filtering using metadata column:", meta_col, "\n")
```

```r
# --- Filter Cells for Target Patient ---
cells_keep <- rownames(meta)[grepl(target, as.character(meta[[meta_col]]), fixed = TRUE)]
if(length(cells_keep) == 0) stop("No cells found for the target patient.")

obj_filt <- subset(obj, cells = cells_keep)
cat("Number of cells selected for", target, 
```

```r
# --- Ensure percent.mt Exists ---
if(!"percent.mt" %in% colnames(obj_filt@meta.data)){
  if(any(grepl("^MT-", rownames(obj_filt)))) obj_filt[["percent.mt"]] <- PercentageFeatureSet(obj_filt, pattern = "^MT-")
  else if(any(grepl("^mt-", rownames(obj_filt)))) obj_filt[["percent.mt"]] <- PercentageFeatureSet(obj_filt, pattern = "^mt-")
  else obj_filt[["percent.mt"]] <- NA
}
```

```r
# --- QC Summary ---
qc_df <- obj_filt@meta.data %>%
  dplyr::select(
    nFeature_RNA = dplyr::any_of("nFeature_RNA"),
    nCount_RNA = dplyr::any_of("nCount_RNA"),
    percent.mt = dplyr::any_of("percent.mt")
  )

summary_stats <- data.frame(
  metric = c("nCells",
             "nFeature_RNA_mean","nFeature_RNA_median",
             "nCount_RNA_mean","nCount_RNA_median",
             "percent.mt_mean","percent.mt_median"),
  value = c(
    ncol(obj_filt),
    mean(qc_df$nFeature_RNA, na.rm = TRUE), median(qc_df$nFeature_RNA, na.rm = TRUE),
    mean(qc_df$nCount_RNA, na.rm = TRUE), median(qc_df$nCount_RNA, na.rm = TRUE),
    mean(qc_df$percent.mt, na.rm = TRUE), median(qc_df$percent.mt, na.rm = TRUE)
  )
)

summary_stats
```

```r
# --- Visualize QC Metrics ---
# Violin plots
VlnPlot(obj_filt, features = c("nFeature_RNA", "nCount_RNA", "percent.mt"), ncol = 1, pt.size = 0.1)
```

```r
# Scatter plot: nCount_RNA vs nFeature_RNA
FeatureScatter(obj_filt, feature1 = "nCount_RNA", feature2 = "nFeature_RNA")
```

```r
# --- Dimensionality Reduction (PCA/UMAP) ---
# Run PCA if not present
if(!"pca" %in% names(obj_filt@reductions)){
  obj_filt <- NormalizeData(obj_filt, verbose = FALSE)
  obj_filt <- FindVariableFeatures(obj_filt, verbose = FALSE)
  obj_filt <- ScaleData(obj_filt, verbose = FALSE)
  obj_filt <- RunPCA(obj_filt, verbose = FALSE)
}

pc_count <- if("pca" %in% names(obj_filt@reductions)) ncol(obj_filt@reductions$pca@cell.embeddings) else 0
use_dims <- seq_len(min(20, pc_count))
```

```r
# Run UMAP if not present
if(!"umap" %in% names(obj_filt@reductions)){
  obj_filt <- FindNeighbors(obj_filt, dims = use_dims, verbose = FALSE)
  obj_filt <- RunUMAP(obj_filt, dims = use_dims, verbose = FALSE)
}

# UMAP Plot
DimPlot(obj_filt, reduction = "umap", group.by = meta_col) + ggtitle(paste0("UMAP colored by ", meta_col))
```

```r
# --- Save Filtered Seurat Object and Summary ---
out_rds <- file.path(outdir, paste0(target, "_seurat_filtered.rds"))
saveRDS(obj_filt, file = out_rds)

write.csv(summary_stats, file = file.path(outdir, paste0("qc_summary_", target, ".csv")), row.names = FALSE)

cat("Filtered Seurat object saved to:", out_rds, "\n")
cat("EDA summary saved to:", outdir, "\n")
```
